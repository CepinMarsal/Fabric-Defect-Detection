```python
from ultralytics import YOLO
from pathlib import Path
```

```python
DATA_YAML = Path("../dataset/balanced dataset/data-balanced.yaml").resolve()

print("Dataset YAML:")
print(DATA_YAML)

print("\nFile exists:", DATA_YAML.exists())
```

```python
DATASET_DIR = DATA_YAML.parent

for split in ["train", "testing", "evaluation"]:
    image_dir = DATASET_DIR/split/"images"
    label_dir = DATASET_DIR/split/"labels"

    print(f"\n{split.upper()}")
    print("Images :", image_dir.exists())
    print("Labels :", label_dir.exists())
    print("Jumlah images :", len(list(image_dir.glob("*"))))
    print("Jumlah labels :", len(list(label_dir.glob("*.txt"))))
```

```python
model = YOLO("yolo11s.pt")
```

```python
test_results = model.train(
    data=str(DATA_YAML),
    epochs=1,
    imgsz=320,
    batch=4,
    device="cpu",
    project="../runs",
    name="yolo11s_test"
)
```

```python
model = YOLO("yolo11s.pt")

results = model.train(
    data=str(DATA_YAML),
    epochs=30,
    imgsz=416,
    batch=16,
    device=0,
    project="../runs",
    name="yolo11s_fabric_defect"
)
```

```python
RUN_DIR = Path("../runs/yolo11s_fabric_defect").resolve()

print("Training result:")
print(RUN_DIR)
print("\nBest model:")
print(RUN_DIR/"weights"/"best.pt")
```

```python
best_model = YOLO(str(RUN_DIR/"weights"/"best.pt"))

metrics = best_model.val(
    data=str(DATA_YAML),
    split="test",
    imgsz=320,
    batch=4,
    device="cpu"
)
```

```python
print("Evaluation Results")
print(f"mAP50    : {metrics.box.map50:.4f}")
print(f"mAP50-95 : {metrics.box.map:.4f}")
```

```python
evaluation_images = DATASET_DIR/"evaluation"/"images"
sample_image = next(evaluation_images.glob("*"))

print("Sample evaluation image:")
print(sample_image)
```

```python
results = best_model.predict(
    source=str(sample_image),
    imgsz=320,
    conf=0.25,
    device="cpu",
    save=True
)
```
